```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("random_stock_market_dataset.csv")
df['Date'] = pd.to_datetime(df['Date'])

fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(18,10),
    gridspec_kw={'height_ratios':[3,1]},
    sharex=True
)

# Price range
ax1.fill_between(df['Date'], df['Low'], df['High'],
                 alpha=0.3, label='Daily Range')

# Price lines
ax1.plot(df['Date'], df['Open'], linewidth=1.5, label='Open')
ax1.plot(df['Date'], df['Close'], linewidth=2.5, label='Close')

ax1.set_title("Stock Market Price Trend")
ax1.set_ylabel("Price")
ax1.legend()
ax1.grid(True)

# Volume
ax2.bar(df['Date'], df['Volume'])
ax2.set_title("Trading Volume")
ax2.set_ylabel("Volume")
ax2.grid(True)

plt.tight_layout()
plt.show()
```
